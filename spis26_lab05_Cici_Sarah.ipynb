```python
from PIL import Image
```

```python
def grayscale(im):
  (width, height) = im.size
  for x in range(width):
    for y in range(height):
      (r,g,b) = im.getpixel((x,y))
      value = int(r*(0.21)) + int(g*(0.72)) + int(b*(0.07))
      im.putpixel((x,y), (value, value, value))

nature = Image.open("nature.jpg")
bear = Image.open("bear.png")
grayscale(nature)
grayscale(bear)
display(nature)
display(bear)
```

```python
def binarize(im, thresh, startx, starty, endx, endy):
  (w, h) = im.size
  if 0 > startx:
    print("ERROR 1")
    return
  elif w < endx:
    print("ERROR 2")
    return
  elif startx > endx:
    print("ERROR 3")
    return
  elif 0 > starty:
    print("ERROR 4")
    return
  elif h < endy:
    print("ERROR 5")
    return
  elif starty > endy:
    print("ERROR 6")
    return
  for x in range(startx, endx):
    for y in range(starty, endy):
      (r,g,b) = im.getpixel((x,y))
      lum = int(r*(0.21)) + int(g*(0.72)) + int(b*(0.07))
      if lum > thresh:
        im.putpixel((x, y), (255, 255, 255))
      else:
        im.putpixel((x, y), (0, 0, 0))

nature = Image.open("nature.jpg")
bear = Image.open("bear.png")
binarize(nature, 127, 0, 0, 1500, 1000)
binarize(bear, 100, 200, 300, 400, 600)
display(nature)
display(bear)
```

```python
def mirrorVert(im):
  (w, h) = im.size
  for x in range(w):
    for y in range(h // 2):
      im.putpixel((x, h - y - 1), (im.getpixel((x, y))))

nature = Image.open("nature.jpg")
bear = Image.open("bear.png")
mirrorVert(nature)
mirrorVert(bear)
display(nature)
display(bear)
```

```python
def mirrorHoriz(im):
  (w, h) = im.size
  for x in range(w // 2):
    for y in range(h):
      im.putpixel((w - x - 1, y), (im.getpixel((x, y))))

nature = Image.open("nature.jpg")
bear = Image.open("bear.png")
mirrorHoriz(nature)
mirrorHoriz(bear)
display(nature)
display(bear)
```
