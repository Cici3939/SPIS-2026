```python
def sumDigits(x):
  sum = 0
  while x > 0:
    sum += x % 10
    x //= 10
  return sum

print(sumDigits(2341234))
```

```python
def getNumber():
  num = 100
  numbers = []
  while num >= 0:
    symbols = input("Enter a digit: ")
    num = int(symbols)
    if num >= 0:
      numbers.append(num)
    else:
      break
  return numbers
print(*getNumber(), sep = "")
```

```python
# returns the wage gap based on provided country and men's wage
def convertWageMtoW(mWage, country):
  if country == "US":
    currWageGap = .179
  elif country == "Korea":
    currWageGap = .278
  elif country == "Japan":
    currWageGap = .201
  else:
    print("Country not recognized. Here is the US gender wage gap.")
    currWageGap = .179
  ratio = 1 - currWageGap
  return mWage * ratio

print(convertWageMtoW(100, "US"))
print(convertWageMtoW(76.2, "Korea"))
print(convertWageMtoW(0, "Japan"))
print(convertWageMtoW(-77.8, "Mexico"))
print(convertWageMtoW(28397491234, "Canada"))
```

```python
# word-guessing game
def wordGuessGame(target, mistakesAllowed):
  target = target.upper()
  partial = ""
  for c in target:
    partial += "*"
  mistakes = 0
  print(partial)
  while mistakes < mistakesAllowed + 1:
    guess = input("Enter a character: ").upper()
    if guess in target:
      for i in range(len(target)):
        if target[i] == guess:
          partial = partial[:i] + guess + partial[i+1:]
    else:
      mistakes += 1
    print(partial)
    print(f"Mistakes: {mistakes}")
    if partial == target:
      print("Yay")
      break
  if mistakes == mistakesAllowed + 1:
    print("oh no")

wordGuessGame("Python", 1)
```

```python
import string

def is_secure(password):

  if len(password) < 7:
    return "insecure"
  else:
    count = 0
    hasUpper = False
    hasLower = False
    hasDigit = False
    hasPunc = False
    for c in password:
      if c.isupper():
        hasUpper = True
      elif c.islower():
        hasLower = True
      elif c.isdigit():
        hasDigit = True
      elif c in string.punctuation:
        hasPunc = True
    if hasUpper:
      count += 1
    if hasLower:
      count += 1
    if hasDigit:
      count += 1
    if hasPunc:
      count += 1
    if (count>2):
      return "secure"
  return "insecure"

print("a", is_secure("a"))
print("iLOVEpython12", is_secure("iLOVEpython12"))
print("OOPsTheBomb", is_secure("OOPsTheBomb"))
```

```python
def rotate(s, n):
  return s[len(s)-n:] + s[:len(s)-n]

print(rotate("Burrow", 0)) #should return "Burrow"

print(rotate("Burrow", 1)) #should return "wBurro"

print(rotate("Burrow", 2)) #should return "owBurr"

print(rotate("Burrow", 3)) #should return "rowBur"

print(rotate("Burrow", 4)) #should return "rrowBu"

print(rotate("Burrow", 5)) #should return "urrowB"
```

```python
def triforce(num):
  # first triangle
  unNum = 1
  numCopy = num
  for i in range((numCopy + 1) // 2):
    for j in range(numCopy):
      print("-", end = "")
    for j in range(unNum):
      print("*", end = "")
    for j in range(numCopy):
      print("-", end = "")
    print()
    unNum += 2
    numCopy -= 1
  # second layer of triangles (2)
  asterisk = 1
  dash = (num - asterisk) // 2
  for i in range((num + 1) // 2):
    # triangle 1
    for j in range(dash):
      print("-", end = "")
    for j in range(asterisk):
      print("*", end = "")
    for j in range(dash):
      print("-", end = "")
    # middle
    print("-", end = "")
    # triangle 2
    for j in range(dash):
      print("-", end = "")
    for j in range(asterisk):
      print("*", end = "")
    for j in range(dash):
      print("-", end = "")
    asterisk += 2
    dash = (num - asterisk) // 2
    print()

triforce(5)
print()
triforce(7)
```

```python
def triforceOrient(h,w,orientation):
  matrix = []

  y = 0
  starLoc = w//2
  numStar = 1
  while y<h//2:
    x=0
    matrix.append([])
    while x < w:
      if x == starLoc:
        for i in range(numStar):
          matrix[y].append("*")
        x+=numStar

      else:
        matrix[y].append("-")
        x+=1

    print()
    starLoc -= 1
    numStar += 2
    y+=1
  y=0
  starLoc = w//4
  numStar = 1

  endInd = len(matrix)

  while y<h//2:
    matrix.append([])
    x2=0
    x=0
    while x < w//2:
      if x == starLoc:
        for j in range(numStar):
          matrix[y+endInd].append("*")
        x+=numStar
      else:
        matrix[y+endInd].append("-")
        x+=1
    while x2 < int((w/2 + 0.5)):
      if x2 == starLoc:
        for j in range(numStar):
          matrix[y+endInd].append("*")
        x2+=numStar
      else:
        matrix[y+endInd].append("-")
        x2+=1

    starLoc -= 1
    numStar += 2
    y+=1
  if orientation=="left":
    matrix = [list(row) for row in zip(*matrix)]
  elif orientation=="down":
    matrix = matrix[::-1]
  elif orientation=="right":
    matrix = matrix[::-1]
    matrix = [list(row) for row in zip(*matrix)]
  for r in matrix:
    print(*r)

triforceOrient(13,23,"right")
```
