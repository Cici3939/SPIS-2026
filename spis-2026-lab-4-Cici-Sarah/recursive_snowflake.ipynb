```python
def snowflake(side_length, levels):
  for i in range (3):
    snowflake_side(side_length,levels)
    left(120)
```

```python
def snowflake_side(side_length, levels):
  if (levels == 0):
    forward(side_length)
  else:
    snowflake_side(side_length/3,levels-1)
    right(60)
    snowflake_side(side_length/3,levels-1)
    left(120)
    snowflake_side(side_length/3,levels-1)
    right(60)
    snowflake_side(side_length/3,levels-1)
```

```python
clearscreen()
speed(0)
snowflake(140,4)
done()
clearscreen()
speed(100)
snowflake(140,0)
done()
```
