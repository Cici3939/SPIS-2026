```python
def tree(trunk_length, height):
  if height == 0:
    return
  else:
    forward(trunk_length)
    left(45)
    tree(trunk_length//1.5,height-1)
    right(90)
    tree(trunk_length//1.5,height-1)
    left(45)

    penup()
    backward(trunk_length)
    pendown()


clearscreen()
left(90)
penup()
setpos(0,-300)
pendown()

#please kindly add one to height for whatever height you expect to see
tree(200,5)
```
