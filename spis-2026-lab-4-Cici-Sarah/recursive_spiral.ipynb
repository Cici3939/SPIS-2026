```python
def spiral(initial_length, angle, multiplier):
  if multiplier > 1 and initial_length > 200:
    return
  elif multiplier < 1 and initial_length < 1:
    return
  else:
    forward(initial_length)
    right(angle)
    spiral(initial_length*multiplier,angle,multiplier)

clearscreen()

spiral(100,90,0.9)

clearscreen()

spiral(1,-45,1.1)
```
