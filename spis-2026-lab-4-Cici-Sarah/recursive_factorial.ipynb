```python
def factorial(n):
    '''Return n! (the product of all integers from 1 up to n).
       factorial(0) is 1.'''
    if n == 0:
     return 1
    else:
      return n * factorial(n - 1)

print(factorial(0))
print(factorial(1))
print(factorial(5))
print(factorial(7))
```
