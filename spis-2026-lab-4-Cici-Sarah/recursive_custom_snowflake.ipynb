```python
def snowflake(side_length, levels, num_sides):
    '''Draw a Koch snowflake built on a regular polygon with
       num_sides sides (num_sides=3 gives the original triangle).'''
    for i in range(num_sides):
      snowflake_side(side_length,levels)
      left(360 / num_sides)
```

```python
def snowflake_side(side_length, levels):
  if (levels == 0):
    forward(side_length)
  else:
    snowflake_side(side_length/3,levels-1)
    right(60)
    snowflake_side(side_length/3,levels-1)
    left(120)
    snowflake_side(side_length/3,levels-1)
    right(60)
    snowflake_side(side_length/3,levels-1)
```

```python
clearscreen()
speed(0)
snowflake(140,4, 3)
done()
clearscreen()
speed(100)
snowflake(120, 3, 4)
done()
clearscreen()
speed(100)
snowflake(100, 3, 5)
done()
```
