```python
def factor(x, y):
  '''Return True if y is a factor of x, and False otherwise.
       Solve this using recursion -- do NOT use the % operator.'''
  if ((x - y) < y):
    if (x - y) != 0:
      return False
    else:
      return True
  else:
    return factor(x-y, y)

print(factor(12, 3))
print(factor(12, 5))
print(factor(20, 4))
print(factor(7, 2))
```
