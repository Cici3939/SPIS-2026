```python
def fibonacci(n):
    '''Return the nth Fibonacci number (0-indexed):
       fibonacci(0) = 0, fibonacci(1) = 1, and each later value
       is the sum of the previous two.'''
    if n == 0:
      return 0
    elif n == 1:
      return 1
    else:
      return fibonacci(n - 1) + fibonacci(n - 2)

print(fibonacci(0))
print(fibonacci(1))
print(fibonacci(4))
print(fibonacci(6))
print(fibonacci(10))
```
