```python
# Precondition: must be an equilateral triangle
def sierpinski(p1, p2, p3, level):
    '''Draw a Sierpinski triangle on the triangle with corners
       p1, p2, p3, recursed down to the given level.'''
    if level == 0:
      setpos(p2)
      pendown()
      setpos(p3)
      setpos(p1)
      setpos(p2)
    else:
      sierpinski(find_midpoint(p1, p2), p2, find_midpoint(p2, p3), level - 1)
      setpos(find_midpoint(p2, p3))
      sierpinski(find_midpoint(p1, p3), find_midpoint(p2, p3), p3, level - 1)
      setpos(find_midpoint(p1, p2))
      sierpinski(p1, find_midpoint(p1, p2), find_midpoint(p1, p3), level - 1)
      setpos(p2)
```

```python
def find_midpoint(p1, p2):
  return ((p1[0] + p2[0]) / 2, (p1[1] + p2[1]) / 2)
```
