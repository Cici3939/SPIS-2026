```python
#draw letter C
def draw_C(the_turtle):

  the_turtle.penup()
  the_turtle.setpos(100,200)
  the_turtle.pendown()

  #the_turtle.backward(200)

  the_turtle.left(135)

  #draw using straight lines
  the_turtle.setpos(100,200)
  the_turtle.forward(100)

  the_turtle.left(45)
  the_turtle.forward(250)

  the_turtle.left(45)
  the_turtle.forward(100)

  the_turtle.left(30)
  the_turtle.forward(100)

  the_turtle.left(15)
  the_turtle.forward(100)

  the_turtle.left(15)
  the_turtle.forward(100)

  the_turtle.left(30)
  the_turtle.forward(100)

  the_turtle.left(45)
  the_turtle.forward(250)

  the_turtle.left(45)
  the_turtle.forward(100)


  #draw using .circle()
  the_turtle.left(90)
  the_turtle.penup()
  the_turtle.setpos(10,100)
  the_turtle.pendown()

  the_turtle.circle(200, 270)


clearscreen()
turt = Turtle()

turt.speed(5)
draw_C(turt)
```
