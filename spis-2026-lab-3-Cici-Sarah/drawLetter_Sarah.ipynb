```python
# draws the letter S
def draw_S(turt):
  """ draw_S(turtle) takes in a turtle parameter to make it draw an S shape"""
  turt.left(90)

  # draw top curve
  for i in range(18):
    turt.forward(5)
    turt.left(10)
  turt.forward(5)

  # draw connection
  for i in range(10):
    turt.left(9)
    turt.forward(5)

  # draw bottom curve
  for i in range(33):
    turt.forward(5)
    turt.right(8)
  turt.forward(5)

clearscreen()

aturtle = Turtle()
bturtle = Turtle()

aturtle.penup()
aturtle.setpos(100, 100)
aturtle.pendown()

draw_S(aturtle)
draw_S(bturtle)

```
