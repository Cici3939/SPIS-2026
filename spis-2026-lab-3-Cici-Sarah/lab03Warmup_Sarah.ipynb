```python
from ColabTurtlePlus.Turtle import *
```

```python
def draw_picture(the_turtle):

    ''' Draw a simple picture using a turtle '''

    the_turtle.forward(50)

    the_turtle.left(45)

    the_turtle.forward(25)

    the_turtle.left(30)

    the_turtle.forward(100)

    the_turtle.left(60)

    the_turtle.forward(150)

    the_turtle.left(90)


clearscreen()

my_turtle = Turtle()      # Create a new Turtle object

my_turtle.speed(5)

draw_picture(my_turtle)   # make the new Turtle draw the shape


turtle1 = Turtle()

turtle2 = Turtle()

turtle1.setpos(-50, -50)

turtle2.setpos(200, 100)

turtle1.forward(100)

turtle2.left(90)

turtle2.forward(100)
```
