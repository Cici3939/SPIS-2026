```python
def draw_C(the_turtle, size):
  """ draw_C(the_turtle, size) takes in a turtle parameter to make it draw a C shape """
  # setup before drawing
  the_turtle.left(135)
  the_turtle.penup()
  the_turtle.setpos(the_turtle.xcor() + (size * 1.8), the_turtle.ycor())
  the_turtle.pendown()

  # draw C
  the_turtle.circle(size, 270)

  # setup to go back to top
  the_turtle.penup()
  the_turtle.circle(size, 90)
  the_turtle.right(135)

def draw_I(the_turtle, size):
  """ draw_I(the_turtle, size) takes in a turtle parameter to make it draw an I shape """
  # setup before drawing
  the_turtle.penup()
  the_turtle.forward(size * 0.4)
  the_turtle.left(90)
  the_turtle.forward(size * 0.2) # change to go more up @ start
  the_turtle.pendown()
  the_turtle.left(180)
  the_turtle.forward(size * 1.8)
  the_turtle.left(180)
  the_turtle.forward(size * 1.8)

  the_turtle.penup()
  the_turtle.setpos(the_turtle.xcor() + (size // 2), the_turtle.ycor() - 0.2 * size)
  the_turtle.right(90)

def draw_letter(the_turtle, letter, size):
  """ calls the draw function on the given letter """
  if letter == "C":
    draw_C(the_turtle, size)
  elif letter == "I":
    draw_I(the_turtle, size)

def draw_word(the_turtle, word, size):
  """ draw the word """
  for c in word:
    draw_letter(the_turtle, c, size)




clearscreen()

aturtle = Turtle()

draw_word(aturtle, "CICI", 50)
aturtle.setpos(aturtle.xcor() - 100, aturtle.ycor() - 100)
draw_word(aturtle, "CICI", 5)
```
